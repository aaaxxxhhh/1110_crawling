```python
import os
import re
from bs4 import BeautifulSoup

# HTML 파일 경로 설정
file_path = '/Users/aaa/Desktop/APP/20261106_이벤트신청페이지_v10_Claude.html'

# Colab 환경에서 로컬 파일 경로를 바로 읽을 수 없는 경우를 대비해 파일 존재 여부 확인
if os.path.exists(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        html_content = f.read()

    # BeautifulSoup을 이용해 스크립트 태그 분석
    soup = BeautifulSoup(html_content, 'html.parser')
    scripts = soup.find_all('script')

    print("--- 발견된 GA / GTM 관련 스크립트 및 이벤트 ---\n")

    # GA, GTM, gtag, dataLayer 관련 키워드 검색
    keywords = ['ga', 'gtag', 'google-analytics', 'googletagmanager', 'dataLayer', 'UA-', 'G-']
    found_any = False

    for i, script in enumerate(scripts):
        script_text = script.string or script.get('src', '')
        if any(key.lower() in script_text.lower() for key in keywords):
            found_any = True
            print(f"[Script #{i+1}]")
            if script.get('src'):
                print(f"  External Source: {script.get('src')}")
            if script.string:
                # 핵심적인 부분 매칭을 위해 출력 제한 또는 전체 출력
                lines = script.string.strip().split('\n')
                relevant_lines = [line for line in lines if any(k in line for k in ['gtag', 'dataLayer', 'ga'])]
                if relevant_lines:
                    print("  Relevant Lines:")
                    for line in relevant_lines:
                        print(f"    {line.strip()}")
                else:
                    print(f"  Content snippet: {script.string.strip()[:300]}...")
            print("-" * 40)

    # 인라인 속성 (onclick 등)에서 ga/gtag 추적 코드 검색
    inline_trackers = []
    for tag in soup.find_all(True):
        for attr in ['onclick', 'onchange', 'href']:
            val = tag.get(attr, '')
            if isinstance(val, str) and any(k in val for k in ['gtag', 'ga', 'dataLayer']):
                inline_trackers.append((tag.name, attr, val))

    if inline_trackers:
        print("\n--- 발견된 인라인 클릭/이벤트 추적 코드 ---")
        for tag_name, attr, val in inline_trackers:
            print(f"<{tag_name} {attr}=\"{val}\">")

    if not found_any and not inline_trackers:
        print("파일 내에서 GA(Google Analytics) 관련 코드나 이벤트를 찾지 못했습니다.")
else:
    print(f"파일을 찾을 수 없습니다: {file_path}\n컴퓨터 내부 로컬 경로이므로, 해당 HTML 파일을 현재 Colab 세션에 업로드한 뒤 실행해야 분석이 가능할 수 있습니다. 업로드 후 경로를 다시 지정해주세요.")
```
